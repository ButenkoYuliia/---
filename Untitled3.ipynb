```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import cm
from matplotlib.colors import Normalize
from sklearn.datasets import make_blobs

# 1. Данные: 4 кластера в 3D
# Центры заданы вручную, чтобы кластеры стояли как на картинке
centers = [[-1, 4.5, 0],   # Cluster 1 — красный, спереди справа
           [4, -3.5, 2],   # Cluster 2 — зелёный, слева
           [-4, 2, 3],     # Cluster 3 — синий, справа
           [-3, -3, 6]]    # Cluster 4 — жёлтый, сзади сверху
X, y = make_blobs(n_samples=500, centers=centers, n_features=3,
                  cluster_std=0.8, random_state=42)
df = pd.DataFrame(X, columns=['Feature 1', 'Feature 2', 'Feature 3'])
df['Cluster'] = y

# Цвета кластеров
colors = ['#ff4b4b', '#4CAF50', '#3498db', '#f1c40f']

# 2. Волнистая поверхность ("холмы") под точками
grid = np.linspace(-6, 6, 120)
GX, GY = np.meshgrid(grid, grid)
Z_raw = (1.5 * np.exp(-((GX + 2) ** 2 + (GY - 1) ** 2) / 6)
         + 1.2 * np.exp(-((GX - 2) ** 2 + (GY + 2) ** 2) / 5)
         - 1.5 * np.exp(-((GX - 3) ** 2 + (GY - 3) ** 2) / 8)
         - 0.8 * np.exp(-((GX + 3) ** 2 + (GY + 3) ** 2) / 6))
Z_raw = np.clip(Z_raw, -1.5, 1.5)
GZ = Z_raw * 3 - 2          # растягиваем по высоте для красоты

cmap = cm.viridis
norm = Normalize(vmin=-1.5, vmax=1.5)

# 3. Тёмная тема и фигура
plt.style.use('dark_background')
fig = plt.figure(figsize=(10, 7))
fig.patch.set_facecolor('#1e1e1e')
ax = fig.add_subplot(111, projection='3d')
ax.set_facecolor('#1e1e1e')
ax.computed_zorder = False   # точки всегда рисуются поверх поверхности

# Поверхность
ax.plot_surface(GX, GY, GZ, facecolors=cmap(norm(Z_raw)),
                rstride=2, cstride=2, linewidth=0.2,
                edgecolor=(1, 1, 1, 0.15), alpha=0.85, shade=True)

# 4. Точки кластеров
for i in range(4):
    part = df[df['Cluster'] == i]
    ax.scatter(part['Feature 1'], part['Feature 2'], part['Feature 3'],
               c=colors[i], s=60, edgecolor='white', linewidth=0.5,
               depthshade=True, label=f'Cluster {i + 1}')

# 5. Цветовая шкала справа
mappable = cm.ScalarMappable(norm=norm, cmap=cmap)
mappable.set_array([])
fig.colorbar(mappable, ax=ax, shrink=0.6, pad=0.08)

# 6. Подписи, сетка, легенда, угол обзора
ax.set_xlabel('Feature 1', fontsize=12, labelpad=10)
ax.set_ylabel('Feature 2', fontsize=12, labelpad=10)
ax.set_zlabel('Feature 3', fontsize=12, labelpad=10)
for axis in (ax.xaxis, ax.yaxis, ax.zaxis):
    axis.set_pane_color((0.12, 0.12, 0.12, 1.0))
    axis._axinfo['grid']['color'] = (1, 1, 1, 0.15)

ax.set_xlim(-6, 6)
ax.set_ylim(-6, 6)
ax.set_zlim(-6, 8)
ax.legend(loc='upper right')
ax.view_init(elev=20, azim=35)
plt.tight_layout()
plt.show()
```

### Interactive 3D Plot with Plotly

Now, let's create an interactive version of the 3D plot using Plotly, which will allow for rotation and zooming.

```python
import plotly.graph_objects as go

# Create scatter traces for each cluster
scatter_traces = []
for i in range(len(centers)):
    part = df[df['Cluster'] == i]
    scatter_traces.append(go.Scatter3d(
        x=part['Feature 1'],
        y=part['Feature 2'],
        z=part['Feature 3'],
        mode='markers',
        marker=dict(
            size=5,
            color=colors[i], # Use the same colors as matplotlib
            line=dict(width=0.5, color='white')
        ),
        name=f'Cluster {i + 1}'
    ))

# Create the surface trace
surface_trace = go.Surface(
    x=GX,
    y=GY,
    z=GZ,
    colorscale='Viridis', # Corresponds to matplotlib's viridis colormap
    opacity=0.85,
    showscale=True, # Show color scale
    cmin=Z_raw.min(), # Set cmin/cmax for color scale to match Z_raw range
    cmax=Z_raw.max(),
    surfacecolor=Z_raw # Use Z_raw for coloring the surface
)

# Combine all traces
fig = go.Figure(data=scatter_traces + [surface_trace])

# Update layout for dark theme and labels
fig.update_layout(
    scene=dict(
        xaxis_title='Feature 1',
        yaxis_title='Feature 2',
        zaxis_title='Feature 3',
        xaxis=dict(
            backgroundcolor="#1e1e1e",
            gridcolor="rgba(255,255,255,0.15)",
            linecolor="rgba(255,255,255,0.15)",
            zerolinecolor="rgba(255,255,255,0.15)"
        ),
        yaxis=dict(
            backgroundcolor="#1e1e1e",
            gridcolor="rgba(255,255,255,0.15)",
            linecolor="rgba(255,255,255,0.15)",
            zerolinecolor="rgba(255,255,255,0.15)"
        ),
        zaxis=dict(
            backgroundcolor="#1e1e1e",
            gridcolor="rgba(255,255,255,0.15)",
            linecolor="rgba(255,255,255,0.15)",
            zerolinecolor="rgba(255,255,255,0.15)"
        )
    ),
    plot_bgcolor='#1e1e1e',
    paper_bgcolor='#1e1e1e',
    font=dict(color='white'),
    title_text='Interactive 3D Cluster Plot with Surface',
    height=700,
    width=1000,
    legend=dict(x=0.9, y=0.9, bgcolor='rgba(0,0,0,0)', bordercolor='rgba(0,0,0,0)')
)

# Display the figure
fig.show()
```
